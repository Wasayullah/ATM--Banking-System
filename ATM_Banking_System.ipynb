# 🏦 ATM Banking System

A beginner-friendly Python project that simulates a simple ATM.

## Features

- PIN authentication
- Check balance
- Deposit money
- Withdraw money
- Change PIN
- Transaction history
- Maximum 3 PIN attempts
- Basic input validation

```python
# Initial account information

correct_pin = "1234"
balance = 10000

# List to store all transactions
transactions = []

print("Account created successfully!")
print(f"Initial Balance: Rs. {balance}")
```

```python
# Allow the user a maximum of 3 attempts to enter the PIN

attempts = 0
login_successful = False

while attempts < 3:

    entered_pin = input("Enter your 4-digit PIN: ")

    if entered_pin == correct_pin:
        login_successful = True
        print("\nLogin successful!")
        break

    else:
        attempts += 1
        remaining_attempts = 3 - attempts

        print("Incorrect PIN!")

        if remaining_attempts > 0:
            print(f"Attempts remaining: {remaining_attempts}")

if not login_successful:
    print("\nToo many incorrect attempts.")
    print("Your account has been temporarily locked.")
```

```python
def check_balance():
    """
    Display the current account balance.
    """

    print("\n========== BALANCE ==========")
    print(f"Current Balance: Rs. {balance}")
    print("=============================")
```

```python
def deposit_money():
    """
    Deposit money into the account.
    """

    global balance

    try:
        amount = float(input("Enter deposit amount: "))

        # Check that the amount is positive
        if amount <= 0:
            print("Please enter an amount greater than 0.")
            return

        balance += amount
        
         # Save transaction
        transactions.append(f"Deposit: Rs. {amount:.2f}")

        print(f"\nRs. {amount:.2f} deposited successfully.")
        print(f"Current Balance: Rs. {balance:.2f}")

    except ValueError:
        print("Invalid amount. Please enter a number.")
```

```python
def withdraw_money():
    """
    Withdraw money from the account.
    """

    global balance

    try:
        amount = float(input("Enter withdrawal amount: "))

        # Check that the amount is positive
        if amount <= 0:
            print("Please enter an amount greater than 0.")
            return

        # Check whether enough money is available
        if amount > balance:
            print("\nInsufficient balance.")
            print(f"Available Balance: Rs. {balance:.2f}")
            return
        
        balance -= amount
        
          # Save transaction
        transactions.append(f"Withdrawal: Rs. {amount:.2f}")

        print(f"\nRs. {amount:.2f} withdrawn successfully.")
        print(f"Remaining Balance: Rs. {balance:.2f}")

    except ValueError:
        print("Invalid amount. Please enter a number.")

        
```

```python
def change_pin():
    """
    Allow the user to change their PIN.
    """

    global correct_pin

    old_pin = input("Enter your current PIN: ")

    # Verify old PIN
    if old_pin != correct_pin:
        print("Incorrect current PIN.")
        return 

    new_pin = input("Enter your new 4-digit PIN: ")

    # Validate PIN length
    if len(new_pin) != 4:
        print("PIN must contain exactly 4 digits.")
        return

    # Validate that PIN contains only numbers
    if not new_pin.isdigit():
        print("PIN must contain only numbers.")
        return

    # Prevent using the same PIN
    if new_pin == correct_pin:
        print("New PIN cannot be the same as the old PIN.")
        return

    correct_pin = new_pin

    print("PIN changed successfully!")    
    
```

```python
def show_transactions():
    """
    Display all transactions made during this session.
    """

    print("\n====== TRANSACTION HISTORY ======")

    if len(transactions) == 0:
        print("No transactions found.")

    else:
        for number, transaction in enumerate(transactions, start=1):
            print(f"{number}. {transaction}")

    print("=================================")
```

```python
def show_menu():
    """
    Display the ATM menu.
    """

    print("\n========== ATM SYSTEM ==========")
    print("1. Check Balance")
    print("2. Deposit Money")
    print("3. Withdraw Money")
    print("4. Change PIN")
    print("5. Transaction History")
    print("6. Exit")
    print("===============================")
```

```python
# Run the ATM only if login was successful

if login_successful:

    while True:

        show_menu()

        choice = input("Enter your choice: ")

        if choice == "1":
            check_balance()

        elif choice == "2":
            deposit_money()

        elif choice == "3":
            withdraw_money()

        elif choice == "4":
            change_pin()

        elif choice == "5":
            show_transactions()

        elif choice == "6":
            print("\nThank you for using our ATM.")
            print("Goodbye!")
            break

        else:
            print("\nInvalid choice.")
            print("Please select a number from 1 to 6.")
```

```python
# Check current balance

check_balance()
```

```python
print("\n========== FINAL ACCOUNT SUMMARY ==========")
print(f"Final Balance: Rs. {balance:.2f}")
print(f"Total Transactions: {len(transactions)}")
print("===========================================")
```
